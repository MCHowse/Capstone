Step 5: Filter NTD Data to Tennessee

```python
# The Agency Information file has a 'State' column
# Filter to Tennessee agencies only
tn_agencies = ntd_agencies[ntd_agencies['State'].astype(str).str.strip() == 'TN'].copy()

print(f"Total agencies in NTD: {len(ntd_agencies)}")
print(f"Tennessee agencies: {len(tn_agencies)}")
print()
print("Tennessee agencies found:")
print(tn_agencies[['NTD ID', 'Agency Name', 'City', 'State']].to_string())
```

```python
# Use the Tennessee NTD IDs to filter the Service data
tn_ntd_ids = tn_agencies['NTD ID'].tolist()

tn_service = ntd_service[ntd_service['NTD ID'].isin(tn_ntd_ids)].copy()

print(f"Total service rows in NTD: {len(ntd_service)}")
print(f"Tennessee service rows: {len(tn_service)}")
print()
tn_service.head()
```

Step 6: Check for Missing Values

```python
print("=== Missing Value Summary ===")
print()
print("B08201 (Vehicle Availability) - missing values per column:")
print (b08201.isnull().sum()[b08201.isnull().sum() > 0])
print()
print("B08301 (Commute Methods) - missing values per column:")
print (b08301.isnull().sum()[b08301.isnull().sum() > 0])
print()
print("B19013 (Median Income) - missing values per column:")
print (b19013.isnull().sum()[b19013.isnull().sum() > 0])
print()
print("S1701 (Poverty) - missing values per column (first 10):")
print (s1701.isnull().sum()[s1701.isnull().sum() > 0].head(10))
print()
print("TN Service - missing values per column:")
print(tn_service.isnull().sum()[tn_service.isnull().sum() > 0])
```
